```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load processed splits and raw class mapping
df_train = pd.read_csv('../data/processed/train_split.csv')
df_test = pd.read_csv('../data/processed/test_split.csv')

# Combine splits for dataset-wide visualization
df_labeled = pd.concat([df_train, df_test], axis=0)

# Plot class counts per time step
plt.figure(figsize=(14, 6))
sns.countplot(data=df_labeled, x='time_step', hue='label', palette={0: 'skyblue', 1: 'salmon'})
plt.title('Licit (0) vs. Illicit (1) Transactions Across Time Steps (1-49)', fontsize=14)
plt.xlabel('Time Step')
plt.ylabel('Transaction Count')
plt.legend(title='Class', labels=['Licit (0)', 'Illicit (1)'])
plt.axvline(x=33.5, color='black', linestyle='--', label='Train/Test Split Boundary')
plt.tight_layout()
plt.show()
```
